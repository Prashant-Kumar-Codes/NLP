# Tokenization in Natural Language Processing (NLP)

## What is Tokenization?
**Tokenization** is the fundamental pre-processing step in NLP. It involves breaking down a large body of text (referred to as a **corpus**) into smaller, manageable units called **tokens**. 

Depending on the granularity, tokenization can be of different types:
1. **Sentence Tokenization (Segmentation)**: Splitting a paragraph/document into individual sentences (e.g., using `sent_tokenize`).
2. **Word Tokenization**: Splitting sentences or paragraphs into individual words, punctuation marks, and symbols.
3. **Subword/Character Tokenization**: Splitting words into smaller units (common in modern transformer models like BERT/GPT).

Tokenization helps convert unstructured text into a structured sequence of tokens, which can then be vectorized (converted to numbers) and fed into machine learning and deep learning models.

```python
corpus = """Hello! this is Prashant Kuamr. Here's my learning code on tokenization.
Tokenization is one of the first topics to learn in NLP. In simple words you can say tokenization means break big language data (paragraphs) smaller to sentences and smaller to words.
"""
print(corpus)
```

```python
# Tokenize corpus(paragraph) to sentence
from nltk.tokenize import sent_tokenize

documents = sent_tokenize(corpus)
documents
```

```python
print(f'Type of documents: {type(documents)}')
```

```python
# Tokeniztion
# Paragraph (corpus) to words
from nltk.tokenize import word_tokenize

para_to_words = word_tokenize(corpus)
para_to_words
```

```python
# Sentence (document) to words
sent_to_words = []
for sent in documents:
    sent_to_words += word_tokenize(sent)

sent_to_words
```

```python
# word tokenization using punkwork
from nltk.tokenize import wordpunct_tokenize
sent_to_words2 = wordpunct_tokenize(corpus)
sent_to_words2
#puntuation are separated better than work_tokenize
```

```python
print(len(sent_to_words))
print(len(sent_to_words2))
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokernizer = TreebankWordTokenizer()
tokernizer.tokenize(corpus)
```

## Comparison of Tokenizers in NLTK

In this notebook, we explored three different word-level tokenizers from NLTK. Here is how they compare based on their behavior and implementation:

| Feature / Tokenizer | `word_tokenize` | `wordpunct_tokenize` | `TreebankWordTokenizer` |
| :--- | :--- | :--- | :--- |
| **Description** | NLTK's default, recommended word tokenizer. | A regex-based tokenizer splitting by word vs. non-word characters. | A tokenizer following Penn Treebank conventions. |
| **Contraction Handling** (e.g., `"Here's"`) | Splits contractions into logical semantic units: `['Here', "'s"]`. | Splits strictly by punctuation boundaries: `['Here', "'", 's']`. | Splits contractions similarly to Penn Treebank rules: `['Here', "'s"]`. |
| **Punctuation Separation** | Separates punctuation cleanly in most contexts. | Separates all punctuation marks completely from words. | Only separates punctuation if it is at the end of the input or space-separated. |
| **Sentence Dependency** | Works well on both sentences and raw paragraphs. | Works on raw paragraphs using pure regex rules. | **Requires sentence-tokenized input.** If run on a paragraph, it fails to split periods in intermediate sentences (e.g., keeps `"Kuamr."` and `"NLP."` intact). |
| **Underlying Method** | Uses an improved Penn Treebank-style tokenizer. | Uses a regular expression: `\w+|[^\w\s]+`. | Uses treebank specific regex rules for contractions. |
