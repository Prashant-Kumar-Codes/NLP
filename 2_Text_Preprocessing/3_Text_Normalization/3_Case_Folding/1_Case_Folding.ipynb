# 3. Case Folding in NLP

Case folding is the process of mapping all characters in text to a uniform case—typically lowercasing—so that word variations like `"NLP"`, `"Nlp"`, and `"nlp"` are matched as the exact same token.

### Difference between `str.lower()` and `str.casefold()`:
- **`str.lower()`:** Standard ASCII/Unicode lowercasing.
- **`str.casefold()`:** Aggressive Unicode case removal designed for internationalization (e.g., German `"ß"` becomes `"ss"`).

### Why Use Case Folding?
- Reduces vocabulary size and matrix sparsity in statistical models (BoW, TF-IDF).
- Ensures case-insensitive keyword searching and matching.

### When NOT to Use Case Folding:
- Named Entity Recognition (NER), where capitalized words often signify proper nouns (e.g., `"Apple"` vs `"apple"`).
- Sentiment Analysis where ALL-CAPS words convey intense emotion (`"GREAT!"` vs `"great"`).

## Step 1: Standard Lowercasing vs Unicode Case Folding

```python
text = "NATURAL LANGUAGE PROCESSING IS GREAT! YAY"

# Standard lowercasing
lower_text = text.lower()

# Unicode casefold
casefold_text = text.casefold()

print("Original Text:   ", text)
print("lower():         ", lower_text)
print("casefold():      ", casefold_text)

# German example where lower() != casefold()
german_text = "Fluß"  # German word for river (contains 'ß')
print("\n--- German Character Comparison ---")
print("German Word: ", german_text)
print("lower():     ", german_text.lower())
print("casefold():  ", german_text.casefold())
```

## Step 2: Selective / Conditional Case Folding

```python
# Preserving acronyms (all-uppercase words length <= 4) while lowercasing rest
text_with_acronyms = "The ISRO saTEllIte colleCTED vaLUable data for INDIAN researchers."

def selective_casefold(text):
    words = text.split()
    processed_words = []
    for word in words:
        # If word is an acronym (all uppercase and length > 1), keep original
        if word.isupper() and len(word) > 1:
            processed_words.append(word)
        else:
            processed_words.append(word.lower())
    return " ".join(processed_words)

print("Original: ", text_with_acronyms)
print("Selective Case Fold: ", selective_casefold(text_with_acronyms)) 
```

## Step 3: Applying Case Folding to Pandas DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "review_id": [1, 2, 3],
    "raw_review": [
        "EXCELLENT Product! WOULD BUY AGAIN.",
        "Average Quality... Not BAD.",
        "WORST EXPERIENCE EVER!"
    ]
})

df["casefold_cleaned_review"] = df["raw_review"].str.casefold()
df.head()
```
