## spaCy Lemmatizer
spaCy provides rule-based lemmatization supported by statistical Part-of-Speech (POS) tagging.
- Unlike NLTK's `WordNetLemmatizer` (which requires manual POS tags), spaCy automatically determines the POS tag for tokens in a sentence context and applies the correct lemma via `token.lemma_`.
- Uses language models like `en_core_web_sm` to process text.

```python
import spacy
# download the spacy model (run once)
# spacy.cli.download('en_core_web_sm')
```

```python
# Load English language pipeline
nlp = spacy.load('en_core_web_sm')
```

```python
words = ['go', 'going','gone','eat', 'ate', 'eating', 'eaten', 'write', 'wrote', 'writing', 'written','programming', 'program', 'programmed','history','story',"he's", "she's"]
```

```python
# Lemmatizing individual words
for word in words:
    doc = nlp(word)
    for token in doc:
        print(f"{token.text} -------> {token.lemma_}")
```

```python
# Context-aware Lemmatization in full sentences (spaCy automatically detects POS tags)
sentence = """Ram was going to fetch some water and started watering the flowers. After he finished with water work he went to the market to hit the hitting game. While he gamming he had to go for a walk with his friend Ramesh. He wanted to spend some time with Ram while Ram though wny do Ramesh is spoiling my game I was already at the finishing line."""
doc = nlp(sentence)

for token in doc:
    print(f"{token.text:<12} POS: {token.pos_:<8} Lemma: {token.lemma_}")
```
