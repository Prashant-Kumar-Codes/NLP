```python
corpus = """Ram was going to fetch some water and started watering the flowers. After he finished with water work he went to the market to hit the hitting game. While he gamming he had to go for a walk with his friend Ramesh. He wanted to spend some time with Ram while Ram though wny do Ramesh is spoiling my game I was already at the finishing line."""
words = ['go', 'going','gone','eat', 'ate', 'eating', 'eaten', 'write', 'wrote', 'writing', 'written','programming', 'program', 'programmed','history','story',"he's", "she's"]
```

## PorterStemmer Class
- Operates on single tokens (use tokenization then PorterStemmer().stem(word)); not designed to process a full corpus string directly.
- Fast and widely used in IR/NLP preprocessing but can over-stem or produce non-dictionary stems.
- For better accuracy consider Snowball/Porter2 or lemmatization when morphological correctness matters.

```python
## PorterStemmer
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmer.stem(corpus)
## stemming does not work directly on the corputs intead works on single word
```

```python
for i in words:
    print(i+'---------->'+stemmer.stem(i))
```

```python
stemmer.stem('congratulations')
```

## RegexpStemmer Class
- Uses user-defined regular expressions (regex patterns) to identify and remove morphological affixes (prefixes or suffixes like `ing$`, `ed$`, `ly$`, etc.).
- Highly customizable but entirely deterministic—it relies strictly on the provided regex patterns without any linguistic rules or dictionary verification.
- Includes a `min` parameter to specify the minimum length of the resulting stem, preventing over-stemming of short words (e.g., preventing "sing" or "ring" from being stripped if `min` is set appropriately).
- Best suited for targeted, domain-specific normalization or removing specific affixes rather than general-purpose English stemming.

```python
from nltk.stem import RegexpStemmer
regexpstemmer = RegexpStemmer("ing$|s$|es%|able$|'s$|ed$", min=4)
```

```python
for word in words:
    print(word+'------>'+regexpstemmer.stem(word))
```

## SnowballStemmer Class
- Also known as the **Porter2 Stemmer**; designed as an improved, faster, and more accurate refinement of the original Porter Stemmer algorithm.
- **Multilingual Support**: Supports stemming across multiple languages (e.g., English, French, German, Spanish, Russian) by passing the language name: `SnowballStemmer('english')`.
- **Improved Logic**: Better handles special prefix/suffix stripping rules and edge cases compared to standard `PorterStemmer`.
- **Best Practice**: Preferred over `PorterStemmer` for modern NLP pipelines when morphological precision or multi-language support is needed.

```python
# Snowball Stemmer
from nltk.stem import SnowballStemmer
snow_stemmer = SnowballStemmer(language='english')
for word in words:
    print(word+'--------->'+snow_stemmer.stem(word))
```
