##### **Goal:** Given a corpus convert in the final form by removing stopwords and applying stemming

```python
corpus="""Earth, our celestial home, holds profound significance for humanity and all life it sustains. At the core of its importance lies its capacity to nurture and sustain life. With its unique combination of atmosphere, water, and diverse ecosystems, Earth provides the essential elements for life to thrive. From the depths of the oceans to the peaks of its highest mountains, Earth teems with a staggering array of life forms, each intricately connected in a delicate web of ecosystems.

Secondly, Earth’s role as a planetary system is crucial. Its orbit around the sun, tilted axis, and rotation cycle give rise to the seasons, shaping climates and influencing life patterns across the globe. This dynamic interplay of astronomical factors dictates everything from weather patterns to the distribution of species.

Moreover, Earth serves as a geological marvel, with its ever-changing surface shaped by processes such as erosion, volcanic activity, and plate tectonics. These geological forces not only sculpt the planet’s terrain but also contribute to the formation of resources essential for human civilization, including minerals, fossil fuels, and fertile soil.

Furthermore, Earth’s interconnectedness with the cosmos highlights its broader significance. As part of the solar system and the wider universe, Earth is subject to cosmic phenomena such as solar radiation, cosmic rays, and gravitational interactions. These cosmic influences not only shape Earth’s environment but also drive scientific inquiry into the broader mysteries of the universe.

Finally, Earth’s status as the only known inhabited planet underscores its uniqueness and the urgency of preserving its delicate balance. As humanity grapples with environmental challenges such as climate change, pollution, and habitat destruction, safeguarding the health of our planet becomes an imperative for the well-being of future generations and the preservation of life as we know it."""
```

```python
## importing the stopwords from the courpus module
from nltk.corpus import stopwords
```

```python
## downlaod the stopwords data first(one time run)
# import nltk
# nltk.download('stopwords')
```

```python
## access all the stopwords
stopwords_list = stopwords.words('english')
stopwords_list
```

```python
print(type(stopwords_list))
```

## Tokenizing

```python
## first let's tokenize the corpus we have
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
## tokenize paragraph to sentences
docs = sent_tokenize(corpus)

## tokenize sentences to words
words = word_tokenize(corpus)
```

```python
print(len(words))
words
```

## Stemming

```python
## import Snowball stemming class and creating its object

from nltk.stem import SnowballStemmer
snow_stem = SnowballStemmer(language='english')
```

```python
## appling stemming on only those words which are not available in stopwords list 
final_corpus = list(filter(lambda word: word not in stopwords_list, words))
final_corpus = ' '.join([snow_stem.stem(word) for word in final_corpus])
```

```python
print(final_corpus)
```

```python
## list of sentence of the corpurs after applying stemming only on those words not in stopwords
final_sent_list = []
for sentence in docs:
    words = word_tokenize(sentence)
    final_sent_list.append(' '.join([snow_stem.stem(word) for word in words if word not in stopwords_list]))
```

```python
final_sent_list
```

## Lemmatizing

```python
## using spacy lemmatization for lemmatizing accurately
import spacy
lemmatizer = spacy.load('en_core_web_sm')

## appling lemmatization on only those words which are not available in stopwords list 

final_lemmetized_sent_list = []

# Lemmatizing individual words
for sentence in docs:
    sent = []
    words = word_tokenize(sentence)
    for word in words:
        if word not in stopwords_list:
            doc = lemmatizer(word)
            sent.append(doc[0].lemma_)
            
    final_lemmetized_sent_list.append(' '.join(sent))
```
