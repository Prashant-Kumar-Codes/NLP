# 1. Reading Text Files (.txt)

Read `.txt` files in Python for Natural Language Processing tasks:
- Reading text files using standard Python built-ins (`open()`, `read()`, `readline()`, `readlines()`)
- Handling text encodings (`utf-8`, `latin-1`, `utf-16`) and encoding errors
- Reading multiple text files from a directory (`os`, `glob`, `pathlib`)
- Memory-efficient line-by-line reading for large text files

```python
import os
from pathlib import Path
```

## Step 1: Create Sample Text Files

```python
# current working directory
cur_dir = Path.cwd()

# create a data folder
data_folder = cur_dir  / 'data_folder'
data_folder.mkdir(parents=True, exist_ok=True)

# Create sample data directory
sample_text_1 = """Natural Language Processing (NLP) is a subfield of artificial intelligence.
It focuses on the interaction between computers and human language.
NLP techniques are used in sentiment analysis, chatbot design, and machine translation."""

sample_text_2 = """Text collection is the first step in any NLP pipeline.
Data can come from files, web pages, databases, or APIs.
Proper encoding handling ensures text is loaded accurately."""

file_path_1 = data_folder / 'sample1.txt'
file_path_2 = data_folder / 'sample2.txt'

with open(file_path_1, "w", encoding="utf-8") as f:
    f.write(sample_text_1)

with open(file_path_2, "w", encoding="utf-8") as f:
    f.write(sample_text_2)

print("Sample text files created successfully!")
```

## Step 2: Reading Single Text Files using `open()`

```python
# 1. Reading entire content into a single string using f.read()
with open(file_path_1, "r", encoding="utf-8") as f:
    content = f.read()

print("--- Entire Content ---")
print(content)
print(f"\nTotal characters: {len(content)}")
```

```python
# 2. Reading file into a list of lines using f.readlines()
with open(file_path_1, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("--- Readlines Output ---")
for i, line in enumerate(lines, 1):
    print(f"[Line {i}]: {line.strip()}") # strip extra spaces and new line (\n) characters
```

## Step 3: Efficient Memory Reading for Large Text Files

```python
# Iterating line by line directly over the file object (Memory Efficient)
with open(file_path_1, "r", encoding="utf-8") as f:
    for line_no, line in enumerate(f, 1):
        print(f"[Line {line_no}] {line.rstrip()}")
```

## Step 4: Handling Different Text Encodings

```python
try:
    with open(file_path_2, mode='r', encoding='utf-8', errors='replace') as f:
        for line_no, line in enumerate(f, 1):
            print(f'[Line_No: {line_no}] {line.strip()}')
except Exception as e:
    print(e)
f.close()
```

```python
# Reading file with potential encoding issues or different character sets
# Common encodings: 'utf-8', 'latin-1' (iso-8859-1), 'utf-16', 'cp1252'

try:
    with open(file_path_2, "r", encoding="utf-8", errors="replace") as f:
        text = f.read()
    print("Successfully read file with utf-8 encoding")
except UnicodeDecodeError as e:
    print(f"Encoding error encountered: {e}")
```

## Step 5: Reading Multiple `.txt` Files in a Folder

```python
# Method 1: Using pathlib (converting string data_folder to Path object)
data_folder_path = Path(data_folder)

corpus_documents = []
for filepath in data_folder_path.glob("*.txt"):
    with open(filepath, "r", encoding="utf-8") as f:
        corpus_documents.append({
            "filename": filepath.name,
            "text": f.read()
        })
        
print(corpus_documents)
print(f"Total text files loaded (using Pathlib): {len(corpus_documents)}")

for doc in corpus_documents:
    print(f"\nFile: {doc['filename']}")
    print(f"Text snippet: {doc['text'][:60]}...")

# Method 2: Using glob.glob with string folder path
corpus_glob = []
for filepath in glob.glob(f"{data_folder}/*.txt"):
    with open(filepath, "r", encoding="utf-8") as f:
        corpus_glob.append({
            "filename": os.path.basename(filepath),
            "text": f.read()
        })

print(f"\nTotal text files loaded (using glob.glob): {len(corpus_glob)}")
```
