# 2. Reading CSV & Excel Files

How to load text datasets stored in `.csv`, `.tsv`, and `.xlsx` (Excel) formats for NLP applications:
- Reading CSV files using standard Python `csv` module
- Reading CSV/TSV files using `pandas` (`pd.read_csv`)
- Reading Excel files using `pandas` (`pd.read_excel`)
- Preprocessing & cleaning text columns (handling missing text, dropping NaN, selecting text fields)

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

## Step 1: Create Sample CSV and Excel Files

```python
sample_dir = Path("sample_data")
sample_dir.mkdir(exist_ok=True)

# Sample NLP dataset
data = {
    "id": [101, 102, 103, 104, 105],
    "label": ["positive", "negative", "neutral", "positive", "negative"],
    "text": [
        "The product quality is absolutely amazing! Highly recommended.",
        "Very disappointing experience, the item arrived broken and late.",
        "The item works as expected. Standard build quality.",
        "Fantastic customer support and very fast delivery!",
        "Poor performance and battery life drains quickly."
    ]
}

df_sample = pd.DataFrame(data)

# Save as CSV
csv_path = sample_dir / "reviews.csv"
df_sample.to_csv(csv_path, index=False)

# Save as TSV (Tab-separated values)
tsv_path = sample_dir / "reviews.tsv"
df_sample.to_csv(tsv_path, sep="\t", index=False)

# Save as Excel
excel_path = sample_dir / "reviews.xlsx"
df_sample.to_excel(excel_path, index=False)

print("Created sample CSV, TSV, and Excel files!")
```

## Step 2: Reading CSV Files using Standard Python `csv` Module

```python
import csv

# Using csv.reader
with open(csv_path, mode="r", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader)
    print(f"Header: {header}")
    for row in reader:
        print(f"ID: {row[0]}, Label: {row[1]}, Text Snippet: {row[2][:30]}...")
```

```python
# Using csv.DictReader for key-value dictionary rows
with open(csv_path, mode="r", encoding="utf-8") as f:
    dict_reader = csv.DictReader(f)
    for row in dict_reader:
        print(f"[ID {row['id']}] ({row['label']}): {row['text']}")
```

## Step 3: Reading CSV / TSV Files using `pandas`

```python
# Reading CSV with Pandas
df_csv = pd.read_csv(csv_path)
print("--- CSV DataFrame Head ---")
print(df_csv.head())

# Reading TSV with Pandas (specifying delimiter='\t')
df_tsv = pd.read_csv(tsv_path, sep="\t")
print("\n--- TSV DataFrame Shape ---")
print(df_tsv.shape)
```

## Step 4: Reading Excel Files using `pandas` (`pd.read_excel`)

```python
# Reading Excel file using openpyxl engine
df_excel = pd.read_excel(excel_path)
print("--- Excel DataFrame ---")
print(df_excel.info())
print("\nText Column:")
print(df_excel["text"])
```

## Step 5: Text Column Extraction & Handling Missing Text

```python
# Example with missing values
df_with_nan = df_csv.copy()
df_with_nan.loc[2, "text"] = np.nan

# Drop rows with missing text values
df_clean = df_with_nan.dropna(subset=["text"])

# Or fill missing values with empty string or placeholder
df_filled = df_with_nan.fillna({"text": ""})

# Extract text column as a Python list of strings for NLP model input
text_corpus = df_clean["text"].tolist()
print("Extracted Corpus List:")
for text in text_corpus:
    print(f"- {text}")
```

```python
# Cleanup temporary files
## method -1
csv_path.unlink(missing_ok=True)
tsv_path.unlink(missing_ok=True)
excel_path.unlink(missing_ok=True)

if sample_dir.exists():
    sample_dir.rmdir()
print("Cleanup completed.")

## method -2
for ext in ['*.csv','*.tsv', '*.xlsx']:
    for files in sample_dir.glob(ext):
        files.unlink(missing_ok=True)
    
if sample_dir.exists():
    sample_dir.rmdir()
print('Successfully deleted the files and folder')

```
