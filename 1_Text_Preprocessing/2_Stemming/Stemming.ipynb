```python
corpus = """Ram was going to fetch some water and started watering the flowers. After he finished with water work he went to the market to hit the hitting game. While he gamming he had to go for a walk with his friend Ramesh. He wanted to spend some time with Ram while Ram though wny do Ramesh is spoiling my game I was already at the finishing line."""
stem_words = ['go', 'going','gone','eat', 'ate', 'eating', 'eaten', 'write', 'wrote', 'writing', 'written','programming', 'program', 'programmed','history','story']
```

```python
## PorterStemmer
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmer.stem(corpus)
```

```python
for i in stem_words:
    print(i+'---------->'+stemmer.stem(i))
```
