## WordNet Lemmatizer
NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. *This
class uses morphy() function to the WordNet CorpusReader class to find a lemma*.

```python
words = ['go', 'going','gone','eat', 'ate', 'eating', 'eaten', 'write', 'wrote', 'writing', 'written','programming', 'program', 'programmed','history','story',"he's", "she's"]
```

```python
# import WordNetLemmatizer
from nltk.stem import WordNetLemmatizer
```

```python
# creating wordnet object
lemmatizer = WordNetLemmatizer()
```

```python
for word in words:
    print(word+'------->'+lemmatizer.lemmatize(word))
```

```python
for word in words:
    print(word+'------->'+lemmatizer.lemmatize(word, pos='v'))
```
